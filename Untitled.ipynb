```python
import pandas as pd
df = pd.read_csv('survey-completed.csv')
df.head()
```

```python
import re
import numpy as np

code_pattern = re.compile(r'^([A-E]\d{1,2})[:.]\s*')
col_map = {}
for col in df.columns:
    m = code_pattern.match(col.strip())
    if m:
        col_map[m.group(1)] = col

print(f"Matched {len(col_map)} codes:", sorted(col_map.keys()))

def likert_to_num(val):
    if pd.isna(val): return np.nan
    m = re.match(r'^\s*(\d+)\s*=', str(val))
    return int(m.group(1)) if m else np.nan

likert_codes = ['B1','B4','B5'] + [f'C{i}' for i in range(1,7)] + [f'D{i}' for i in range(1,12)] + [f'E{i}' for i in range(1,6)]
for code in likert_codes:
    if code in col_map:
        df[code] = df[col_map[code]].apply(likert_to_num)

df[likert_codes].describe().T[['count','mean','std']]
```

```python
df['Block_Frequency_Score']    = df[['B1','B4','B5']].mean(axis=1, skipna=True)
df['Time_Pressure_Score']      = df[[f'C{i}' for i in range(1,7)]].mean(axis=1, skipna=True)
df['Shortcut_Frequency_Score'] = df[[f'D{i}' for i in range(1,12)]].mean(axis=1, skipna=True)
df['Awareness_Score']          = df[[f'E{i}' for i in range(1,6)]].mean(axis=1, skipna=True)

df[['Block_Frequency_Score','Time_Pressure_Score','Shortcut_Frequency_Score','Awareness_Score']].describe()
```

```python
def cronbach_alpha(items_df):
    items_df = items_df.dropna()
    item_vars = items_df.var(axis=0, ddof=1)
    total_var = items_df.sum(axis=1).var(ddof=1)
    n = items_df.shape[1]
    return (n / (n - 1)) * (1 - item_vars.sum() / total_var)

print("Block (B1,B4,B5):   ", round(cronbach_alpha(df[['B1','B4','B5']]), 3))
print("Time Pressure (C1-6):", round(cronbach_alpha(df[[f'C{i}' for i in range(1,7)]]), 3))
print("Shortcuts (D1-11):  ", round(cronbach_alpha(df[[f'D{i}' for i in range(1,12)]]), 3))
print("Awareness (E1-5):   ", round(cronbach_alpha(df[[f'E{i}' for i in range(1,6)]]), 3))
```

```python
from scipy import stats

def report_corr(x, y, label):
    valid = df[[x, y]].dropna()
    r, pval = stats.pearsonr(valid[x], valid[y])
    print(f"{label}\n  r = {r:.3f}, p = {pval:.4f}, n = {len(valid)}\n")

report_corr('Block_Frequency_Score', 'Shortcut_Frequency_Score', 'H1: Block -> Shortcuts')
report_corr('Shortcut_Frequency_Score', 'Awareness_Score', 'H2: Shortcuts -> Awareness')
report_corr('Awareness_Score', 'E3', 'H3: Awareness -> Personal incident (E3)')
report_corr('Time_Pressure_Score', 'Block_Frequency_Score', 'RQ2: Time pressure -> Block')
report_corr('Time_Pressure_Score', 'Shortcut_Frequency_Score', 'RQ2/RQ3: Time pressure -> Shortcuts')
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

navy = '#1a3a5c'
blue = '#3d6fa8'
orange = '#c1652f'
darkred = '#8a3a2f'

# --- Print correlation + p-value table for H1, H2, H3 ---
def report_corr(x, y, label):
    valid = df[[x, y]].dropna()
    r, p = stats.pearsonr(valid[x], valid[y])
    rho, p_s = stats.spearmanr(valid[x], valid[y])
    sig = "significant" if p < 0.05 else "not significant"
    print(f"{label}")
    print(f"  Pearson  r = {r:.3f}, p = {p:.4f}  ({sig}), n = {len(valid)}")
    print(f"  Spearman rho = {rho:.3f}, p = {p_s:.4f}")
    print()
    return r, p

print("="*65)
print("HYPOTHESIS TESTING \u2014 CORRELATION ANALYSIS")
print("="*65)
r1, p1 = report_corr('Block_Frequency_Score', 'Shortcut_Frequency_Score', 'H1: Block Frequency -> Shortcut Frequency')
r2, p2 = report_corr('Shortcut_Frequency_Score', 'Awareness_Score', 'H2: Shortcut Frequency -> Awareness')
r3, p3 = report_corr('Awareness_Score', 'E3', 'H3: Awareness -> Personal Incident (E3)')

# --- Build the figure ---
fig, axes = plt.subplots(2, 2, figsize=(13, 10))
fig.suptitle("Figure 5 \u2014 Hypothesis Testing: Correlation Analysis (H1\u2013H3)",
             fontsize=16, fontweight='bold', color=navy, y=0.98)

# Panel 1: Correlation heatmap of all four composites + E3
ax = axes[0,0]
corr_cols = ['Block_Frequency_Score','Time_Pressure_Score','Shortcut_Frequency_Score','Awareness_Score','E3']
corr_labels = ['Block\nFreq.','Time\nPressure','Shortcut\nFreq.','Awareness','E3\n(Incident)']
corr_matrix = df[corr_cols].corr()
im = ax.imshow(corr_matrix.values, cmap='RdBu_r', vmin=-1, vmax=1)
ax.set_xticks(range(len(corr_labels))); ax.set_xticklabels(corr_labels, fontsize=9)
ax.set_yticks(range(len(corr_labels))); ax.set_yticklabels(corr_labels, fontsize=9)
for i in range(len(corr_labels)):
    for j in range(len(corr_labels)):
        ax.text(j, i, f"{corr_matrix.values[i,j]:.2f}", ha='center', va='center',
                color='white' if abs(corr_matrix.values[i,j]) > 0.5 else 'black', fontsize=9)
ax.set_title("Correlation Matrix (All Key Variables)", fontweight='bold')
plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

# Panel 2: H1 scatter
ax = axes[0,1]
valid = df[['Block_Frequency_Score','Shortcut_Frequency_Score']].dropna()
ax.scatter(valid['Block_Frequency_Score'], valid['Shortcut_Frequency_Score'], color=navy, alpha=0.6, edgecolor='white', s=60)
z = np.polyfit(valid['Block_Frequency_Score'], valid['Shortcut_Frequency_Score'], 1)
xs = np.linspace(valid['Block_Frequency_Score'].min(), valid['Block_Frequency_Score'].max(), 50)
ax.plot(xs, np.polyval(z, xs), color=orange, linestyle='--', linewidth=2)
ax.text(0.05, 0.92, f"r = {r1:.3f}, p = {p1:.4f}", transform=ax.transAxes, fontweight='bold', color=navy)
ax.set_title("H1: Block Frequency \u2192 Shortcut Frequency", fontweight='bold', fontsize=11)
ax.set_xlabel("Block Frequency Score"); ax.set_ylabel("Shortcut Frequency Score")

# Panel 3: H2 scatter
ax = axes[1,0]
valid = df[['Shortcut_Frequency_Score','Awareness_Score']].dropna()
ax.scatter(valid['Shortcut_Frequency_Score'], valid['Awareness_Score'], color=blue, alpha=0.6, edgecolor='white', s=60)
z = np.polyfit(valid['Shortcut_Frequency_Score'], valid['Awareness_Score'], 1)
xs = np.linspace(valid['Shortcut_Frequency_Score'].min(), valid['Shortcut_Frequency_Score'].max(), 50)
ax.plot(xs, np.polyval(z, xs), color=orange, linestyle='--', linewidth=2)
ax.text(0.05, 0.92, f"r = {r2:.3f}, p = {p2:.4f}", transform=ax.transAxes, fontweight='bold', color=navy)
ax.set_title("H2: Shortcut Frequency \u2192 Awareness", fontweight='bold', fontsize=11)
ax.set_xlabel("Shortcut Frequency Score"); ax.set_ylabel("Awareness Score")

# Panel 4: H3 scatter
ax = axes[1,1]
valid = df[['Awareness_Score','E3']].dropna()
ax.scatter(valid['Awareness_Score'], valid['E3'], color=darkred, alpha=0.6, edgecolor='white', s=60)
z = np.polyfit(valid['Awareness_Score'], valid['E3'], 1)
xs = np.linspace(valid['Awareness_Score'].min(), valid['Awareness_Score'].max(), 50)
ax.plot(xs, np.polyval(z, xs), color=navy, linestyle='--', linewidth=2)
ax.text(0.05, 0.92, f"r = {r3:.3f}, p = {p3:.4f}", transform=ax.transAxes, fontweight='bold', color=navy)
ax.set_title("H3: Awareness \u2192 Personal Incident (E3)", fontweight='bold', fontsize=11)
ax.set_xlabel("Awareness Score"); ax.set_ylabel("E3 (Personal Incident)")

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig('fig_hypothesis_correlations.png', dpi=150, facecolor='white')
plt.show()
```
